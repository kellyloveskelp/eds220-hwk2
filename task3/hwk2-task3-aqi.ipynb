```python
import os
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt

# Read in the data
aqi_17 = pd.read_csv(
    '/Users/kellyanne/Desktop/meds2027/eds220/eds220-hwk2/data/daily_aqi_by_county_2017.zip', compression='zip')
aqi_18 = pd.read_csv(
    '/Users/kellyanne/Desktop/meds2027/eds220/eds220-hwk2/data/daily_aqi_by_county_2018.zip', compression='zip')
# Starting from "/data" not working...

# Combine dataframes into one
aqi = pd.concat([aqi_17, aqi_18])

# Force column text to lower case and replace spaces with underscores
aqi.columns = (aqi.columns
               .str.lower()
               .str.replace(' ','_'))

# Filter datasret for Santa Barbara County and drop unneccessary columns
aqi_sb = aqi[aqi["county_name"] == "Santa Barbara"].drop(
    columns = ["state_name", "county_name", "state_code", "county_code"])

# Reassign date data type from object to datetime
aqi_sb['date'] = aqi_sb['date'].astype('datetime64[ns]')

# Reset index to date and sort descending
aqi_sb = aqi_sb.set_index('date').sort_index()

# Create five day average with rolling() function
aqi_sb['five_day_average'] = aqi_sb['aqi'].rolling('5D').mean()

# Plot it!
fig, ax = plt.subplots()

aqi_sb.reset_index().plot.line(
    x='date', 
    y='aqi', 
    ax=ax, 
    label = "AQI",
    color='lightblue'
    )
aqi_sb.reset_index().plot.line(
    x='date', 
    y='five_day_average', 
    ax=ax, 
    label = "Five Day Average", 
    color='red', 
    alpha = 0.7, 
    linestyle='--'
    )

# Add a line marking the Dec 2017 Thomas Fire
ax.axvline("2017-12-05", label = "Thomas Fire", color='orange', linewidth = 2, alpha = 0.5)

# Update axes
ax.set_title('AQI in Santa Barbara from 2017-2018')
ax.set_xlabel('Date')
ax.set_ylabel('AQI')
ax.legend()
```

In the next cell we will calculate an average over a [rolling window](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.rolling.html) using the `rolling()` method for `pandas.Series`:

- `rolling()` is a lazy method, so we need to specify what we want to calculate over each window before it does something. 
- in this example we use the aggregator function `mean()` to calculate the average over each window
- the parameter '5D' indicates we want the window for our rolling average to be 5 days. 
- we get a `pandas.Series` as output

Store your answer in the `rolling_average` variable. Your result should be a `pandas.Series` indexed by `date`, with the 5-day averages as its values. 

```python
aqi_sb = aqi_sb.sort_index()
```
